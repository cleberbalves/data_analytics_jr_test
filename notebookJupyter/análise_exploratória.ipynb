```python
import os
import pandas as pd

# Caminhos para os arquivos CSV
escolas_path = r'C:\Users\Usuário\Documents\Desafio\data_analytics_jr_test\Data\Data\Data\Escolas'
educandos_path = r'C:\Users\Usuário\Documents\Desafio\data_analytics_jr_test\Data\Data\Data\Perfil dos educandos'

# Carregar arquivos CSV
escolas_df = pd.concat([pd.read_csv(os.path.join(escolas_path, file), delimiter=';', encoding='latin1') for file in os.listdir(escolas_path) if file.endswith('.csv')], ignore_index=True)
educandos_df = pd.concat([pd.read_csv(os.path.join(educandos_path, file), delimiter=';', encoding='latin1') for file in os.listdir(educandos_path) if file.endswith('.csv')], ignore_index=True)

print("Dados carregados com sucesso!")
```

```python
# Visualizar as primeiras linhas dos dados
print(escolas_df.head())
print(educandos_df.head())
```

```python
# Verificar valores faltantes
print("Valores faltantes em escolas_df:")
print(escolas_df.isnull().sum())

print("Valores faltantes em educandos_df:")
print(educandos_df.isnull().sum())

# Verificar duplicatas
print("Duplicatas em escolas_df:", escolas_df.duplicated().sum())
print("Duplicatas em educandos_df:", educandos_df.duplicated().sum())

# Verificar tipos de dados
print("Tipos de dados em escolas_df:")
print(escolas_df.dtypes)

print("Tipos de dados em educandos_df:")
print(educandos_df.dtypes)
```

```python
Proporção entre Alunos pela Perspectiva dos Perfis
```

```python
import matplotlib.pyplot as plt

# Exemplo de análise: Proporção de alunos por sexo
sexo_proporcao = educandos_df['SEXO'].value_counts(normalize=True) * 100
sexo_proporcao.plot(kind='bar', title='Proporção de Alunos por Sexo')
plt.ylabel('Porcentagem')
plt.xlabel('Sexo')
plt.show()
```

```python
# Exemplo de análise: Densidade de alunos por escola
alunos_por_escola = educandos_df.groupby('NOMESC').size()
alunos_por_escola.plot(kind='hist', title='Densidade de Alunos por Escola')
plt.xlabel('Número de Alunos')
plt.ylabel('Frequência')
plt.show()
```

```python
# Exemplo de análise: Proporção de alunos por distrito
alunos_por_distrito = educandos_df['DISTRITO'].value_counts(normalize=True) * 100
alunos_por_distrito.plot(kind='bar', title='Proporção de Alunos por Distrito')
plt.ylabel('Porcentagem')
plt.xlabel('Distrito')
plt.show()
```

```python
# Criar uma nova coluna de faixa etária
educandos_df['FAIXA_ETARIA'] = pd.cut(educandos_df['IDADE'], bins=[0, 5, 10, 15, 20], labels=['0-5', '6-10', '11-15', '16-20'])

# Plotar a distribuição de alunos por faixa etária
faixa_etaria_distribuicao = educandos_df['FAIXA_ETARIA'].value_counts()
faixa_etaria_distribuicao.plot(kind='bar', title='Distribuição de Alunos por Faixa Etária')
plt.ylabel('Número de Alunos')
plt.xlabel('Faixa Etária')
plt.show()
```

```python
# Plotar a proporção de alunos por modalidade de ensino
modalidade_proporcao = educandos_df['MODAL'].value_counts(normalize=True) * 100
modalidade_proporcao.plot(kind='pie', title='Proporção de Alunos por Modalidade de Ensino', autopct='%1.1f%%')
plt.ylabel('')
plt.show()
```
